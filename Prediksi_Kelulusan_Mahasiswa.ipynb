```python
#label library sckit-learn untuk mengubah teks kategorik menjadi angka
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df = df.copy()
kategori_cols = ['jenis_kelamin', 'status_menikah', 'status_akademik']

for col in kategori_cols:
    df [col] = le.fit_transform(df[col])
df
```

```python
#analisis korelasi antar fitur numerik
#Correlation Fitur Numerik
import seaborn as sns
import matplotlib.pyplot as plt

numeric_df = df.select_dtypes(include=['float64', 'int64'])

# Hitung korelasi
correlation_matrix = numeric_df.corr()

# Plot heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, fmt=".2f", cmap='coolwarm', center=0)
plt.title("Correlation Pada Fitur Numerik")
plt.tight_layout()
plt.show()
```

```python
#Normalisasi Data Menggunakan Z-Score
from sklearn.preprocessing import StandardScaler #atau standar devisiasi
import pandas as pd

numerical_cols = ['umur', 'kehadiran_%', 'partisipasi_diskusi_skor',
                  'nilai_tugas_rata-rata', 'aktivitas_e-learning_skor', 'ipk']

# Ganti koma menjadi titik dan ubah menjadi float untuk kolom 'ipk'
df['ipk'] = df['ipk'].astype(str).str.replace(',', '.', regex=False).astype(float)

scaler = StandardScaler()
df[numerical_cols] = scaler.fit_transform(df[numerical_cols])

df
```

```python
x = df.drop(columns=['status_akademik'])
y = df['status_akademik']
df
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=42)
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report
import pandas as pd

knn_model = KNeighborsClassifier(n_neighbors=1)
knn_model.fit(x_train, y_train)
```

```python
#Latih Model KNN
from sklearn.neighbors import KNeighborsClassifier
import numpy as np
from sklearn.metrics import accuracy_score

error_rate_knn = []
k_value = []
pred_knn = []
acc_knn = []
k_range = range(1, 21)

for i in k_range:
    print('Proses Neigboor', i)
    k_value.append(i)
    knn = KNeighborsClassifier(n_neighbors=i)
    knn.fit(x_train, y_train)
    pred_i = knn.predict(x_test)
    pred_knn.append(pred_i)
    error_rate_knn.append(np.mean(pred_i != y_test))
    acc_knn.append(accuracy_score(y_test, pred_i))
    print('Done...')
```

```python
#Tanpa PCA
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 8))

# Grafik Error Rate
plt.subplot(2, 1, 1)
plt.plot(k_value, error_rate_knn, color='blue', linestyle='dashed', marker='o', markerfacecolor='red')
plt.title('Error Rate vs. K Value')
plt.xlabel('K')
plt.ylabel('Error Rate')
plt.grid(True)

# Grafik Accuracy
plt.subplot(2, 1, 2)
plt.plot(k_value, acc_knn, color='blue', linestyle='dashed', marker='o', markerfacecolor='red')
plt.title('Accuracy vs. K Value')
plt.xlabel('K')
plt.ylabel('Accuracy')
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
#Menentukan jumlah tetangga terbaik 
best_k_index = np.argmax(acc_knn)
best_k = k_value[best_k_index]
best_prediction_knn = pred_knn[best_k_index]

print("Best K:", best_k)
```

```python
#Menentukan jumlah fold terbaik saat validasi silang
from sklearn.model_selection import cross_val_score

print("Hasil cross-validation:")
for i in range(2, 10):
    scores = cross_val_score(KNeighborsClassifier(n_neighbors=best_k), x, y, cv=i)
    print(f"> folds={i}, accuracy={scores.mean():.5f}")
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

# Buat confusion matrix
cm = confusion_matrix(y_test, best_prediction_knn)

# Tampilkan confusion matrix sebagai gambar
disp = ConfusionMatrixDisplay(confusion_matrix=cm)
disp.plot(cmap='Blues')
plt.title(f'Confusion Matrix (k={best_k})')
plt.show()
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score

accuracy_neigh = accuracy_score(y_test, best_prediction_knn)
precision_neigh = precision_score(y_test, best_prediction_knn, average='weighted')
recall_neigh = recall_score(y_test, best_prediction_knn, average='weighted')
f1_score_neigh = f1_score(y_test, best_prediction_knn, average='weighted')

print('Accuracy     : ', round(accuracy_neigh*100, 2), '%')
print('Precision    : ', round(precision_neigh*100, 2), '%')
print('Recall       : ', round(recall_neigh*100, 2), '%')
print("F1-score     : ", round(f1_score_neigh*100, 2), '%')
```

```python
from sklearn.svm import SVC

error_rate_svm = []
c_value = []
pred_svm = []
acc_svm = []
c_range = [0.1, 0.5, 1, 5, 10]

for c in c_range:
    print('Proses SVM dengan C =', c)
    c_value.append(c)
    svm = SVC(C=c)
    svm.fit(x_train, y_train)
    pred_i = svm.predict(x_test)
    pred_svm.append(pred_i)
    error_rate_svm.append(np.mean(pred_i != y_test))
    acc_svm.append(accuracy_score(y_test, pred_i))
    print('Done...')
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 8))

# Grafik Error Rate SVM
plt.subplot(2, 1, 1)
plt.plot(c_value, error_rate_svm, color='green', linestyle='dashed', marker='s', markerfacecolor='orange')
plt.title('Error Rate vs. C Value (SVM)')
plt.xlabel('C')
plt.ylabel('Error Rate')
plt.grid(True)

# Grafik Accuracy SVM
plt.subplot(2, 1, 2)
plt.plot(c_value, acc_svm, color='green', linestyle='dashed', marker='s', markerfacecolor='orange')
plt.title('Accuracy vs. C Value (SVM)')
plt.xlabel('C')
plt.ylabel('Accuracy')
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
import numpy as np

best_svm_index = np.argmax(acc_svm)
best_c = c_value[best_svm_index]

from sklearn.svm import SVC

print("\nHasil cross-validation SVM:")
for i in range(2, 10):
    scores = cross_val_score(SVC(C=best_c), x, y, cv=i)
    print(f"> folds={i}, accuracy={scores.mean():.5f}")
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt
best_prediction_svm = pred_svm[np.argmax(acc_svm)]

# Buat confusion matrix SVM
cm_svm = confusion_matrix(y_test, best_prediction_svm)

# Tampilkan confusion matrix
disp_svm = ConfusionMatrixDisplay(confusion_matrix=cm_svm)
disp_svm.plot(cmap='Greens')
plt.title(f'Confusion Matrix SVM (C={best_c})')
plt.show()
```

```python
# === Evaluasi SVM ===
accuracy_svm = accuracy_score(y_test, best_prediction_svm)
precision_svm = precision_score(y_test, best_prediction_svm, average='weighted')
recall_svm = recall_score(y_test, best_prediction_svm, average='weighted')
f1_score_svm = f1_score(y_test, best_prediction_svm, average='weighted')

print('\n=== SVM ===')
print('Accuracy     :', round(accuracy_svm * 100, 2), '%')
print('Precision    :', round(precision_svm * 100, 2), '%')
print('Recall       :', round(recall_svm * 100, 2), '%')
print('F1-score     :', round(f1_score_svm * 100, 2), '%')
```

```python
import os
import joblib
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import MinMaxScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier

# Misalnya kamu sudah punya x_train dan y_train dari data yang sudah di-preprocess (termasuk label encoding)

# Buat pipeline untuk KNN
pipeline_knn = Pipeline([
    ("scaler", MinMaxScaler()),
    ("knn", KNeighborsClassifier(n_neighbors=5))
])

# Buat pipeline untuk SVM
pipeline_svm = Pipeline([
    ("scaler", MinMaxScaler()),
    ("svm", SVC(kernel='rbf', C=1.0, gamma='scale', probability=True))
])

# Buat pipeline untuk Random Forest
pipeline_rf = Pipeline([
    ("scaler", MinMaxScaler()),
    ("rf", RandomForestClassifier(n_estimators=100, random_state=42))
])

# Training model
pipeline_knn.fit(x_train, y_train)
pipeline_svm.fit(x_train, y_train)
pipeline_rf.fit(x_train, y_train)

# Simpan ke folder models/
os.makedirs("models", exist_ok=True)
joblib.dump(pipeline_knn, "models/knn_model.pkl")
joblib.dump(pipeline_svm, "models/svm_model.pkl")
joblib.dump(pipeline_rf, "models/rf_model.pkl")

print(" Semua model berhasil dilatih dan disimpan ke folder 'models/'")
```
